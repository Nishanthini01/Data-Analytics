```python
rfm_df = pd.merge(
    orders,
    order_items,
    on="order_id",
    how="inner"
)

rfm_df.head()
```

```python
rfm_df["order_date"] = pd.to_datetime(rfm_df["order_date"])
```

```python
reference_date = rfm_df["order_date"].max() + pd.Timedelta(days=1)
print(reference_date)
```

```python
print(orders.columns)
```

```python
print(order_items.columns)
```

```python
rfm_data = rfm_df.groupby("customer_id").agg({
    "order_date": lambda x: (reference_date - x.max()).days,
    "order_id": "nunique",
    "total_price": "sum"
}).reset_index()
```

```python
rfm_data.columns = ["customer_id", "Recency", "Frequency", "Monetary"]
```

```python
rfm_data.head()
```

```python
rfm_data.head()
```

```python
def segment_customer(row):
    if row["Monetary"] >= rfm_data["Monetary"].quantile(0.75):
        return "High Value"
    elif row["Monetary"] >= rfm_data["Monetary"].quantile(0.50):
        return "Regular"
    else:
        return "Low Value"

rfm_data["Customer_Segment"] = rfm_data.apply(segment_customer, axis=1)
```

```python
rfm_data.head()
```

```python
rfm_data["Customer_Segment"].value_counts()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=rfm_data, x="Customer_Segment")
plt.title("Customer Segments")
plt.xlabel("Segment")
plt.ylabel("Number of Customers")
plt.show()
```

```python
rfm_data.to_csv("customer_segments.csv", index=False)
```
